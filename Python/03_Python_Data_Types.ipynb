# 3. Python Data Types

Python provides several built-in data types for storing and working with different kinds of data.

In this notebook, we will explain and demonstrate:

1. Integer (`int`)
2. Float (`float`)
3. Complex (`complex`)
4. String (`str`)
5. List (`list`)
6. Tuple (`tuple`)
7. Set (`set`)
8. Dictionary (`dict`)
9. Boolean (`bool`)

We will also identify whether each data type is **mutable** or **immutable**.

### Mutable vs Immutable

- **Mutable** objects can be changed after they are created.
- **Immutable** objects cannot be changed after they are created.

## 1. Integer (`int`)

An integer represents a whole number without a decimal point.

**Mutability: Immutable**

```python
matches_won = 20
matches_lost = 5
points = 60

print("Matches won:", matches_won)
print("Matches lost:", matches_lost)
print("Points:", points)
print("Type:", type(points))
```

## 2. Float (`float`)

A float represents a number containing a decimal point.

**Mutability: Immutable**

```python
player_rating = 8.7
transfer_fee = 75.5
average_goals = 1.75

print("Player rating:", player_rating)
print("Transfer fee:", transfer_fee, "million")
print("Average goals:", average_goals)
print("Type:", type(player_rating))
```

## 3. Complex (`complex`)

A complex number contains a real part and an imaginary part. Python uses `j` for the imaginary part.

Example: `3 + 4j`

**Mutability: Immutable**

```python
number = 3 + 4j

print("Complex number:", number)
print("Real part:", number.real)
print("Imaginary part:", number.imag)
print("Type:", type(number))
```

## 4. String (`str`)

A string is a sequence of characters enclosed in quotes. Strings are used for text such as names and messages.

**Mutability: Immutable**

A string's characters cannot be changed directly after the string is created.

```python
club = "Manchester United"
stadium = "Old Trafford"

print("Club:", club)
print("Stadium:", stadium)
print("Type:", type(club))

print("Uppercase:", club.upper())
print("Number of characters:", len(club))
print("First character:", club[0])
```

## 5. List (`list`)

A list is an ordered collection of items. It can contain different data types and duplicate values.

Lists are created using square brackets `[]`.

**Mutability: Mutable**

Items can be added, removed, or changed.

```python
players = [
    "Bruno Fernandes",
    "Casemiro",
    "Mason Mount"
]

print("Players:", players)
print("First player:", players[0])
print("Type:", type(players))
```

```python
# Lists are mutable

players = ["Bruno Fernandes", "Casemiro", "Mason Mount"]

players.append("Kobbie Mainoo")
players[1] = "Lisandro Martinez"

print("Updated players:", players)
```

## 6. Tuple (`tuple`)

A tuple is an ordered collection similar to a list. Tuples are created using parentheses `()`.

**Mutability: Immutable**

Once created, tuple elements cannot be changed, added, or removed.

```python
club_details = (
    "Manchester United",
    "England",
    "Old Trafford"
)

print("Club details:", club_details)
print("Club name:", club_details[0])
print("Type:", type(club_details))
```

```python
# Demonstrating tuple immutability

club_details = ("Manchester United", "England", "Old Trafford")

try:
    club_details[0] = "Manchester City"
except TypeError as error:
    print("Error:", error)
```

## 7. Set (`set`)

A set is a collection of **unique** elements. Duplicate values are automatically removed.

Sets are created using curly braces `{}` or the `set()` constructor.

**Mutability: Mutable**

Elements can be added to or removed from a set.

```python
trophies = {
    "Premier League",
    "FA Cup",
    "Champions League",
    "FA Cup"
}

print("Trophies:", trophies)
print("Type:", type(trophies))
print("Notice that the duplicate FA Cup is stored only once.")
```

```python
trophies = {"Premier League", "FA Cup"}

trophies.add("Champions League")

print("Updated trophies:", trophies)
```

## 8. Dictionary (`dict`)

A dictionary stores data as **key-value pairs**. A key is used to access its corresponding value.

Dictionaries are created using curly braces `{}`.

**Mutability: Mutable**

Values can be changed, and new key-value pairs can be added or removed.

```python
player = {
    "name": "Bruno Fernandes",
    "club": "Manchester United",
    "position": "Midfielder",
    "shirt_number": 8
}

print("Player:", player)
print("Name:", player["name"])
print("Position:", player["position"])
print("Type:", type(player))
```

```python
player: dict[str, str | int] = {
    "name": "Bruno Fernandes",
    "club": "Manchester United",
    "position": "Midfielder"
}

player["shirt_number"] = 8
player["position"] = "Captain and Midfielder"

print("Updated player:", player)
```

## 9. Boolean (`bool`)

A Boolean represents one of two truth values:

- `True`
- `False`

Booleans are commonly used in conditions and decision-making.

**Mutability: Immutable**

```python
is_manchester_united_fan = True
is_manchester_united_based_in_london = False

print("Is a Manchester United fan:", is_manchester_united_fan)
print("Is Manchester United based in London:", is_manchester_united_based_in_london)
print("Type:", type(is_manchester_united_fan))
```

```python
points = 75
qualified_for_competition = points >= 70

if qualified_for_competition:
    print("The team has qualified for the Champions League.")
else:
    print("The team has not qualified for the Champions League.")
```

## Summary of Python Data Types

| Data Type | Example | Mutable / Immutable |
|---|---|---|
| `int` | `20` | Immutable |
| `float` | `8.7` | Immutable |
| `complex` | `3 + 4j` | Immutable |
| `str` | `"Manchester United"` | Immutable |
| `list` | `["Bruno", "Casemiro"]` | Mutable |
| `tuple` | `("Manchester United", "England")` | Immutable |
| `set` | `{"FA Cup", "Premier League"}` | Mutable |
| `dict` | `{"name": "Bruno"}` | Mutable |
| `bool` | `True` | Immutable |

### Easy Way to Remember

**Immutable:** `int`, `float`, `complex`, `str`, `tuple`, `bool`

**Mutable:** `list`, `set`, `dict`

## Demonstrating Multiple Data Types Together

Real-world programs usually use several data types together. For example, a dictionary can contain strings, integers, floats, lists, booleans, and complex numbers.

```python
player = {
    "name": "Bruno Fernandes",                 # str
    "shirt_number": 8,                         # int
    "rating": 8.7,                             # float
    "is_captain": True,                        # bool
    "positions": ["Midfielder", "Attacking Midfielder"],  # list
    "coordinates": 3 + 4j                     # complex
}

print(player)

print("\nData types:")
for key, value in player.items():
    print(f"{key}: {type(value).__name__}")
```

## Conclusion

Python's built-in data types allow us to represent different kinds of information.

- **Numbers:** `int`, `float`, `complex`
- **Text:** `str`
- **Collections:** `list`, `tuple`, `set`, `dict`
- **Logical values:** `bool`

Understanding **mutability** is important because it determines whether an existing object can be changed after it has been created.
