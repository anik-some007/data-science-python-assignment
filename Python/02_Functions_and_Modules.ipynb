## 2(a). Functions, Parameters, Arguments and Return Values

A **function** is a reusable block of code designed to perform a particular task.

A function can:
1. Accept input through **parameters**
2. Perform some operation
3. Send a result back using a **return value**

### Basic Syntax

```python
def function_name(parameter):
    # code
    return result
```

- **Parameter**: A variable listed when defining a function.
- **Argument**: The actual value passed when calling the function.
- **Return value**: The value sent back by the function using `return`.

````python
def greet_club(club_name):
    return "Welcome, " + club_name + "!"

message = greet_club("Manchester United")
print(message)
````

### Parameters vs Arguments

In the example above:

```python
def greet_club(club_name):
```

`club_name` is the **parameter**.

When we write:

```python
greet_club("Manchester United")
```

`"Manchester United"` is the **argument**.

The function returns a string, which is its **return value**.

````python
def calculate_points(wins, draws, losses):
    points = (wins * 3) + (draws * 1)
    return points

manchester_united_points = calculate_points(20, 5, 3)
print("Manchester United points:", manchester_united_points)
````

## 2(b). Different Types of Function Arguments

### 1. Default Arguments

A default argument provides a value automatically if the caller does not provide one.

````python
def player_info(player, position="Forward"):
    return f"{player} plays as a {position}."

print(player_info("Marcus Rashford"))
print(player_info("Bruno Fernandes", "Midfielder"))
````

### 2. Keyword Arguments

Keyword arguments are passed using the parameter name.

This makes function calls easier to understand.

````python
def match_result(opponent, goals_for, goals_against):
    return f"Manchester United {goals_for} - {goals_against} {opponent}"

result = match_result(
    opponent="Liverpool",
    goals_for=3,
    goals_against=1
)

print(result)
````

### 3. `*args`

`*args` allows a function to accept any number of positional arguments.

Inside the function, `args` is treated as a tuple.

````python
def add_goals(*goals):
    return sum(goals)

total_goals = add_goals(2, 1, 3, 2, 4, 5)
print("Total goals:", total_goals)
````

### 4. `**kwargs`

`**kwargs` allows a function to accept any number of keyword arguments.

Inside the function, `kwargs` is treated as a dictionary.

````python
def player_details(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

player_details(
    name="Bruno Fernandes",
    club="Manchester United",
    position="Midfielder",
    nationality="Portugal"
)
````

### Combining `*args` and `**kwargs`

A function can accept both forms when needed.

````python
def team_summary(*players, **team_info):
    print("Players:")
    for player in players:
        print("-", player)

    print("\nTeam Information:")
    for key, value in team_info.items():
        print(f"{key}: {value}")

team_summary(
    "Bruno Fernandes",
    "Benjamin Sesko",
    "Kobie Mainoo",
    Club="Manchester United",
    League="Premier League"
)
````

## 2(c). Writing Useful Functions

### 1. Find the Largest of Three Numbers

This function compares three numbers and returns the largest one.

````python
def largest_of_three(a, b, c):
    return max(a, b, c)

print("Largest number:", largest_of_three(10, 25, 18))
````

### 2. Check Whether a Number is Prime

A **prime number** is a number greater than 1 that has exactly two factors: 1 and itself.

````python
def is_prime(number):
    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True

number = 29

if is_prime(number):
    print(number, "is a prime number.")
else:
    print(number, "is not a prime number.")
````

### 3. Calculate Factorial

The factorial of a positive integer `n` is the product of all positive integers from `1` to `n`.

For example:

`5! = 5 × 4 × 3 × 2 × 1 = 120`

````python
def factorial(number):
    if number < 0:
        return "Factorial is not defined for negative numbers."

    result = 1

    for i in range(1, number + 1):
        result *= i

    return result

print("5! =", factorial(5))
print("7! =", factorial(7))
````

### 4. Calculate Student Percentage

This function calculates percentage from marks obtained and total marks.

**Formula:**

Percentage = `(marks obtained / total marks) × 100`

````python
def calculate_percentage(marks_obtained, total_marks):
    if total_marks <= 0:
        return "Total marks must be greater than zero."

    return (marks_obtained / total_marks) * 100

percentage = calculate_percentage(425, 500)
print(f"Student percentage: {percentage:.2f}%")
````

### 5. Convert Celsius to Fahrenheit

**Formula:**

`Fahrenheit = (Celsius × 9/5) + 32`

````python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

temperature = 25
print(f"{temperature}°C = {celsius_to_fahrenheit(temperature):.2f}°F")
````

## 2(d). Creating a Custom Module

A **module** is a Python file containing reusable code such as functions, variables, or classes.

We will create a module named:

```text
student_utils.py
```

It will contain five useful functions:

1. `calculate_percentage()`
2. `calculate_grade()`
3. `is_passed()`
4. `calculate_average()`
5. `student_summary()`

The module can then be imported into another Python program.

````python
# Create student_utils.py

module_code = '''
def calculate_percentage(marks_obtained, total_marks):
    if total_marks <= 0:
        return 0
    return (marks_obtained / total_marks) * 100


def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


def is_passed(percentage):
    return percentage >= 40


def calculate_average(marks):
    if not marks:
        return 0
    return sum(marks) / len(marks)


def student_summary(name, marks):
    total_marks = len(marks) * 100
    marks_obtained = sum(marks)
    percentage = calculate_percentage(marks_obtained, total_marks)
    grade = calculate_grade(percentage)

    return {
        "name": name,
        "percentage": percentage,
        "grade": grade,
        "passed": is_passed(percentage)
    }
'''

with open("student_utils.py", "w") as file:
    file.write(module_code)

print("student_utils.py created successfully.")
````

### Importing and Using the Custom Module

Now we can import the functions from `student_utils.py` just like functions from Python's built-in modules.

````python
import student_utils

marks = [85, 78, 92, 88, 80]

percentage = student_utils.calculate_percentage(sum(marks), len(marks) * 100)
grade = student_utils.calculate_grade(percentage)
passed = student_utils.is_passed(percentage)
average = student_utils.calculate_average(marks)

print("Student: Anik Some")
print(f"Average marks: {average:.2f}")
print(f"Percentage: {percentage:.2f}%")
print(f"Grade: {grade}")
print(f"Passed: {passed}")
````

### Using `student_summary()`

We can also use the fifth function to generate a complete student summary.

````python
summary = student_utils.student_summary(
    "Anik Some",
    [85, 78, 92, 88, 80]
)

for key, value in summary.items():
    print(f"{key.title()}: {value}")
````
