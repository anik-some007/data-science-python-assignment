## 1. What is Git?

Git is a distributed version control system used to track changes in files and source code.

Git allows developers to:

- Track the history of a project
- Create and manage versions
- Work safely on changes
- Revert to earlier versions
- Collaborate with other developers

### What is GitHub?

GitHub is a cloud-based platform that hosts Git repositories.

Git and GitHub work together, but they are not the same thing:

- **Git** manages version history locally.
- **GitHub** stores and shares Git repositories online.

## 3. Initializing a Git Repository

The `git init` command initializes a new Git repository in the current project directory.

### Command

```bash
git init
```

Git creates a hidden `.git` directory containing the information required to track the repository's history.

## 4. Checking Repository Status

The `git status` command shows the current state of the working directory and staging area.

### Command

```bash
git status
```

It can show:

- The current branch
- Files that have been modified
- Untracked files
- Files staged for commit
- Whether the working tree is clean

### Example

```text
On branch main
nothing to commit, working tree clean
```

## 5. Staging Files with `git add`

Before committing changes, files are added to the staging area.

### Command

```bash
git add README.md .gitignore
```

Multiple files can also be staged using:

```bash
git add .
```

After staging, `git status` shows the files under:

```text
Changes to be committed
```

## 6. Creating a Commit

A commit records a set of changes in the Git repository's history.

### Command Used

```bash
git commit -m "initial repository setup"
```

The `-m` option provides a message describing the changes.

Example commit created for this project:

```text
initial repository setup
```

## 7. Connecting the Local Repository to GitHub

The local project was connected to the GitHub repository using the remote named `origin`.

### Command

```bash
git remote -v
```

The repository remote used for this assignment is:

```text
https://github.com/anik-some007/data-science-python-assignment.git
```

The `origin` remote is used for both fetching and pushing changes.

## 8. Pushing Changes to GitHub

The `git push` command uploads local commits to a remote GitHub repository.

### Command Used

```bash
git push -u origin main
```

The `-u` option sets `origin/main` as the upstream branch for the local `main` branch.

After this command was executed successfully, the local commit became available in the GitHub repository.

## 9. Checking the Repository After Push

After pushing, `git status` can be used to confirm that the local branch is synchronized with the remote repository.

### Command

```bash
git status
```

Example:

```text
On branch main
Your branch is up to date with 'origin/main'.

nothing to commit, working tree clean
```

This indicates that there are no local changes waiting to be committed or pushed.

## 10. Pulling Changes from GitHub

The `git pull` command downloads changes from the remote repository and integrates them into the local branch.

### Command Used

```bash
git pull origin main
```

For this assignment, the command returned:

```text
From https://github.com/anik-some007/data-science-python-assignment
 * branch            main       -> FETCH_HEAD
Already up to date.
```

This confirmed that the local repository already contained the latest changes from GitHub.
