## 4. Creating a Virtual Environment

A virtual environment creates an isolated Python environment for a project.

This is useful because different projects may require different package versions. Installing packages inside a virtual environment prevents those dependencies from unnecessarily affecting other Python projects.

### Creating the Virtual Environment

From the project directory:

```bash
python -m venv .venv
```

This creates a virtual environment named `.venv`.

Typical project structure after creation:

```text
data-science-python-assignment/
│
├── .venv/
├── Python/
├── EDA/
├── Web_Scraping/
├── Dataset/
└── README.md
```

The `.venv` directory should normally be excluded from Git using `.gitignore`.

## 5. Activating the Virtual Environment on Windows

For Git Bash, the virtual environment can be activated using:

```bash
source .venv/Scripts/activate
```

For Windows PowerShell, the command is:

```powershell
.venv\Scripts\Activate.ps1
```

For Command Prompt:

```cmd
.venv\Scripts\activate.bat
```

After activation, the terminal usually displays `(.venv)` before the command prompt.

````python
# Verify that the notebook is running with the expected Python interpreter.

import sys

print("Python executable:")
print(sys.executable)

print("\nPython version:")
print(sys.version)
````

## 6. Checking `pip` Version

The installed version of pip can be checked with:

```bash
python -m pip --version
```

This confirms that pip is available for the selected Python installation.

````python
# Check the installed pip version from Python.

import subprocess
import sys

result = subprocess.run(
    [sys.executable, "-m", "pip", "--version"],
    capture_output=True,
    text=True
)

print(result.stdout.strip())
````

## 7. Environment Setup Summary
